# Luxury Housing Sales Analysis - Beginner Notebook

Run each cell in order. Use the raw downloaded CSV for cleaning. The source does not contain booking outcomes. MySQL loading is optional and runs only when you execute the final cell.


```python
from pathlib import Path
import pandas as pd
from luxury_housing_analysis import clean_data, load_to_mysql
from housing_visualizations import create_charts
from housing_eda import run_eda
```

## 1. Load the raw CSV
Enter the file path instead of editing a fixed path in the notebook.


```python
csv_path = Path(input('Raw CSV file path: ').strip().strip(chr(34)))
raw_data = pd.read_csv(csv_path)
print('Rows and columns:', raw_data.shape)
raw_data.head()
```

## 2. Inspect structure and missing values
These checks help us understand the data before changing it.


```python
raw_data.info()
raw_data.isna().sum()
```

```python
print('Exact duplicate rows:', raw_data.duplicated().sum())
raw_data.describe(include='all')
```

## 3. Clean and create features
The function removes duplicates, converts prices/dates, handles nulls, creates calendar and Indian fiscal quarters, and preserves missing booking outcomes.


```python
cleaned_data = clean_data(raw_data)
cleaned_data.head()
```

```python
cleaned_data[['Ticket_Price_Cr', 'Unit_Size_Sqft', 'Price_per_Sqft']].isna().sum()
```

## 4. Save the cleaned data
Choose an output folder different from the raw data location. Existing named output files are regenerated.


```python
output_folder = Path(input('Output folder: ').strip().strip(chr(34)))
output_folder.mkdir(parents=True, exist_ok=True)
cleaned_path = output_folder / 'luxury_housing_cleaned.csv'
if cleaned_path.resolve() == csv_path.resolve():
    raise ValueError('Choose another output folder to preserve the raw CSV.')
cleaned_data.to_csv(cleaned_path, index=False)
print('Saved:', cleaned_path.resolve())
```

## 5. Create four simple charts
`groupby` groups rows; `size` counts them; `sum` adds available ticket prices.


```python
figure = create_charts(cleaned_data, output_folder)
figure
```

## 6. Run the fuller EDA
This writes statistics, cross-tabs, price outlier records, correlations, and six charts. Observed-score comparisons exclude median-filled amenity values.


```python
eda_summary = run_eda(cleaned_data, output_folder)
eda_summary
```

## 7. Explain the dataset limitation
Record counts are not booking counts. No genuine booking status is present in the course-linked CSV.


```python
print('Known booking outcomes:', cleaned_data['Booking_Flag'].notna().sum())
print('Booking labels:', cleaned_data['Booking_Status'].value_counts())
cleaned_data.groupby('Fiscal_Quarter')['Property_ID'].count()
```
